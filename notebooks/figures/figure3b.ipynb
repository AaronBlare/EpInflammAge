```python
# Import all necessary packages
import pandas as pd
import numpy as np
from wordcloud import WordCloud
from collections import Counter
import rootutils

# Set directories and load data
path_root = str(rootutils.find_root(indicator=".project-root"))
path_data = f"{path_root}/data/cytokines-regression"
path_plots = f"{path_root}/plots"
data = pd.read_excel(f"{path_data}/union_gsea.xlsx", index_col=0)

data['Keywords'].replace('', np.nan, inplace=True)
data.dropna(subset=['Keywords'], inplace=True)
keywords = np.concatenate(data['Keywords'].str.split(';').values)
keywords_counter = Counter(keywords)
df_keywords_counter = pd.DataFrame.from_dict(keywords_counter, orient='index', columns=['Count'])
df_keywords_counter.sort_values(['Count'], ascending=[False], inplace=True)
keywords_counter_dict = dict(keywords_counter)

# Plot Figure 3b
# Different runs lead to different words colors and positions. No seed was set
wordcloud = WordCloud(width=1600, height=600, background_color='white', colormap='Set1', max_words=400, max_font_size=200).fit_words(keywords_counter_dict)
wordcloud.to_file(f'{path_plots}/figure3b.png')
wordcloud_svg = wordcloud.to_svg(embed_font=True)
with open(f'{path_plots}/figure3b.svg', "w", encoding="utf-8") as svg:
   svg.write(wordcloud_svg)
```
